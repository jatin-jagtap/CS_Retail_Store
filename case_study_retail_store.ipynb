```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
sales_df = pd.DataFrame({
 "OrderID": [1001, 1002, 1003, 1004, 1005, 1006],
 "Product": ["Notebook", "Pen", "Bag", "Notebook", "Bottle", "Pen"],
 "Quantity": [3, 10, 2, 4, 3, 8],
 "UnitPrice": [50, 10, 600, 50, 150, 10],
 "City": ["Indore", "Bhopal", "Indore", "Ujjain", "Bhopal", "Indore"]
})
products_df = pd.DataFrame({
 "Product": ["Notebook", "Pen", "Bag", "Bottle"],
 "Category": ["Stationery", "Stationery", "Accessories", "Accessories"],
 "Stock": [40, 100, 12, 25],
 "CostPrice": [30, 4, 420, 90]
})
customers_df = pd.DataFrame({
 "CustomerID": [1, 2, 3, 4, 5],
 "Customer": ["Asha", "Rahul", "Neha", "Aman", "Ira"],
 "City": ["Indore", "Bhopal", "Indore", "Ujjain", "Dewas"],
 "Age": [24, 31, 28, 22, 35]
})
monthly_sales_df = pd.DataFrame({
 "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
 "Revenue": [12000, 15000, 13500, 18000, 21000, 19500]
})
employee_df = pd.DataFrame({
 "Employee": ["Karan", "Pooja", "Sameer", "Divya"],
 "Department": ["Sales", "Support", "Sales", "Inventory"],
 "MonthlySalary": [25000, 22000, 28000, 24000]
})
```

```python
print(sales_df.shape)
print(products_df.shape)
print(customers_df.shape)
print(monthly_sales_df.shape)
print(employee_df.shape)
```

```python
#Q2
sales_df["Revenue"]=sales_df["Quantity"]*sales_df["UnitPrice"]
sales_df[["OrderID","Product","Quantity","Revenue"]]
```

```python
#Q3
sales_df[sales_df['Revenue'] >=500].sort_values("Revenue",ascending=False)
```

```python
from numpy._core.fromnumeric import sort
#Q4
products_df["Revenue"]=products_df["Stock"]*products_df["CostPrice"]
products_df.max()
```

```python
#Q5
sales_df.groupby('City')['Revenue'].sum().sort_values(ascending=False)
```

```python
#Q6
products_df[["Product","Stock"]][products_df["Stock"] <30]
```

```python
#Q7
merged_df=pd.merge(sales_df, products_df, on='Product')
merged_df["Total_Revenue"]=merged_df["Revenue_x"]+merged_df['Revenue_y']
merged_df["Profit"]=(merged_df["UnitPrice"]-merged_df['CostPrice'])*merged_df["Quantity"]
merged_df[["Product","Quantity","Total_Revenue","Profit"]]
```

```python
#Q8
arr = monthly_sales_df['Revenue'].to_numpy()
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
print(np.sum(arr))
```

```python
#Q9
plt.plot(
    monthly_sales_df["Month"],
    monthly_sales_df["Revenue"],
    marker="o"
)
plt.title("Month VS Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid()
plt.show()
```

```python
#Q10
product_revenue = sales_df.groupby('Product')['Revenue'].sum();
product_revenue.plot(kind='bar',ylabel="Revenue")
plt.title("Revenue Vs Product")
```
